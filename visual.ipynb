```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

```python
df = pd.read_csv("clean_data.csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
print("First 5 rows:")
display(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())
```

```python
print("Data Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum().sum())

print("\nDuplicated Rows:")
print(df.duplicated().sum())
```

```python
df.describe().T
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(df["age"], bins=20, kde=True)

plt.title("Distribution of Player Age")
plt.xlabel("Age")
plt.ylabel("Number of Players")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(df["overall"], bins=20, kde=True)

plt.title("Distribution of Player Ratings")
plt.xlabel("Overall Rating")
plt.ylabel("Number of Players")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(df["potential"], bins=20, kde=True)

plt.title("Distribution of Player Potential")
plt.xlabel("Potential")
plt.ylabel("Number of Players")

plt.show()
```

```python
df["value_million"] = df["value_eur"] / 1_000_000

plt.figure(figsize=(10, 6))

sns.histplot(df["value_million"], bins=30, kde=True)

plt.title("Distribution of Player Market Value")
plt.xlabel("Market Value (€ Million)")
plt.ylabel("Number of Players")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="overall",
    y="potential",
    alpha=0.4
)

plt.title("Overall Rating vs Player Potential")
plt.xlabel("Overall Rating")
plt.ylabel("Potential")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="age",
    y="overall",
    alpha=0.4
)

plt.title("Age vs Player Rating")
plt.xlabel("Age")
plt.ylabel("Overall Rating")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x="overall",
    y="value_million",
    alpha=0.4
)

plt.title("Player Rating vs Market Value")
plt.xlabel("Overall Rating")
plt.ylabel("Market Value (€ Million)")

plt.show()
```

```python
main_features = [
    "pace",
    "shooting",
    "passing",
    "dribbling",
    "defending",
    "physic"
]

for feature in main_features:
    plt.figure(figsize=(8, 5))

    sns.scatterplot(
        data=df,
        x=feature,
        y="overall",
        alpha=0.4
    )

    plt.title(f"{feature.capitalize()} vs Player Rating")
    plt.xlabel(feature.capitalize())
    plt.ylabel("Overall Rating")

    plt.show()
```

```python
correlation_with_overall = (
    df.corr(numeric_only=True)["overall"]
    .drop("overall")
    .sort_values(ascending=False)
)

print("Features most positively correlated with Overall Rating:")
display(correlation_with_overall.head(10))
```

```python
key_features = [
    "overall",
    "potential",
    "age",
    "pace",
    "shooting",
    "passing",
    "dribbling",
    "defending",
    "physic",
    "movement_reactions",
    "skill_ball_control",
    "mentality_composure",
    "value_eur"
]

plt.figure(figsize=(12, 9))

sns.heatmap(
    df[key_features].corr(),
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Heatmap of Key Player Features")

plt.show()
```

```python
correlations = (
    df.corr(numeric_only=True)["overall"]
    .drop("overall")
    .abs()
    .sort_values(ascending=False)
)

print("Top 10 Features Related to Player Rating:")
display(correlations.head(10))
```

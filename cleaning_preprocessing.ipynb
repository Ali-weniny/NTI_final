```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

df = pd.read_csv('Data/FC26_20250921.csv')
print("Shape:", df.shape)
df.head()
```

```python
cols_to_drop = ['work_rate', 'nation_position', 'nation_team_id', 'nation_jersey_number',
                 'player_tags', 'goalkeeping_speed', 'player_traits', 'release_clause_eur']
df = df.drop(columns=cols_to_drop)

df['is_goalkeeper'] = df['player_positions'].str.contains('GK').astype(int)
df = df[df['is_goalkeeper'] == 0].copy()
df = df.drop(columns=['is_goalkeeper'])


gk_cols = ['goalkeeping_diving', 'goalkeeping_handling', 'goalkeeping_kicking',
           'goalkeeping_positioning', 'goalkeeping_reflexes']
df = df.drop(columns=gk_cols)

main_stats = ['pace', 'shooting', 'passing', 'dribbling', 'defending', 'physic']
df[main_stats] = df[main_stats].fillna(0)


df['club_joined_date'] = pd.to_datetime(df['club_joined_date'], errors='coerce')
df['years_at_club'] = (pd.Timestamp('2025-09-19') - df['club_joined_date']).dt.days / 365
df['years_at_club'] = df['years_at_club'].fillna(df['years_at_club'].median())
df = df.drop(columns=['club_joined_date'])

df['contract_years_remaining'] = df['club_contract_valid_until_year'] - 2025
df = df.drop(columns=['club_contract_valid_until_year'])

df = df.dropna(subset=['club_name'])

df['is_loaned'] = df['club_loaned_from'].notna().astype(int)
df = df.drop(columns=['club_loaned_from'])

extra_cols_to_drop = ['player_url', 'long_name', 'dob', 'real_face', 'player_face_url',
                       'body_type', 'club_position',
                       'ls','st','rs','lw','lf','cf','rf','rw','lam','cam','ram','lm',
                       'lcm','cm','rcm','rm','lwb','ldm','cdm','rdm','rwb','lb','lcb','cb','rcb','rb','gk']
df = df.drop(columns=extra_cols_to_drop)

id_cols_to_drop = ['player_id', 'fifa_version', 'fifa_update', 'fifa_update_date',
                    'short_name', 'league_id', 'club_team_id', 'nationality_id',
                    'club_jersey_number']
df = df.drop(columns=id_cols_to_drop)

df = df.drop(columns=['club_name'])


print(df.shape)
null_counts = df.isnull().sum()
```

```python
df.dtypes
```

```python
def group_position(pos):
    pos = pos.split(',')[0].strip()
    if pos in ['CB', 'LB', 'RB', 'LWB', 'RWB']:
        return 'DEF'
    elif pos in ['CDM', 'CM', 'CAM', 'LM', 'RM']:
        return 'MID'
    elif pos in ['ST', 'CF', 'LW', 'RW']:
        return 'ATT'
    else:
        return 'OTHER'

df['position_group'] = df['player_positions'].apply(group_position)
df = df.drop(columns=['player_positions'])
print(df['position_group'].value_counts())
```

```python
print(df['league_name'].nunique())
print(df['nationality_name'].nunique())
```

```python
nationality_freq = df['nationality_name'].value_counts()
df['nationality_freq'] = df['nationality_name'].map(nationality_freq)
df = df.drop(columns=['nationality_name'])
```

```python
league_freq = df['league_name'].value_counts()
df['league_freq'] = df['league_name'].map(league_freq)
df = df.drop(columns=['league_name'])
```

```python
df = pd.get_dummies(df, columns=['preferred_foot', 'position_group'], drop_first=True)
```

```python
print(df.shape)
print(df.dtypes.value_counts())
```

```python
os.makedirs('Data', exist_ok=True)
df.to_csv('Data/clean_data.csv', index=False)
```
